```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine('postgresql://postgres:Manu%408786@localhost:5432/f1_analytics_db')

race_results = pd.read_csv('../data/race_results.csv')
lap_times = pd.read_csv('../data/lap_times.csv')
pit_stops = pd.read_csv('../data/pit_stops.csv')

race_results.to_sql('race_results', engine, if_exists='replace', index=False)
lap_times.to_sql('lap_times', engine, if_exists='replace', index=False)
pit_stops.to_sql('pit_stops', engine, if_exists='replace', index=False)

print("All tables loaded into PostgreSQL successfully!")
```
